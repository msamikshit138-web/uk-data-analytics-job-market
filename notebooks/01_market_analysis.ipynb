# UK Data Analytics Job Market — Exploratory Analysis

This notebook reads the outputs created by the reproducible Python pipeline.  
Run `python src/run_pipeline.py --public` before executing the public-data cells.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

OUTPUTS = Path("../outputs")
```

## UK monthly job-posting trend

```python
trend = pd.read_csv(OUTPUTS / "uk_market_trend_monthly.csv", parse_dates=["date"])
trend.tail()
```

```python
ax = trend.plot(x="date", y="uk_job_postings_index", legend=False)
ax.set_title("UK Job Postings Index")
ax.set_xlabel("Date")
ax.set_ylabel("Index")
plt.show()
```

## Latest UK regional index

```python
regions = pd.read_csv(OUTPUTS / "latest_region_index.csv", parse_dates=["date"])
regions.sort_values("indeed_job_postings_index", ascending=False).head(15)
```
